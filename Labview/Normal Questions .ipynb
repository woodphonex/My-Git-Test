# Palindrome

```python
# Palindrome - 01

def palindrome(string):
    if string == string[::-1]:
        return True
    else:
        return False

print(palindrome("madam"))
```

```python
# Palindrome - 02

def palindrome(string):
    if string == string[::-1]:
        print(string, "=", string[::-1])
        return True
    else:
        print(string, "!=", string[::-1])
        return False

word = str(input("Enter the word :"))
print(palindrome(word))
```

```python
# Palindrome - 03

string = "shas"
if string == string[::-1]:
    print(True)
else:
    print(False)
```

```python
# Palindrome - 04

list_n = ["shas", "bhuv", "level", "sahiti", "madam"]
list_new = []

for i in list_n:
    if i == i[::-1]:
        list_new.append("Palindrome")
    else:
        list_new.append("Not a Palindrome")
print(list_new)
```

# Fibonacci Series

### Each new number is the sum of the previous two.

```python
# Fibonacci Series - 01

n = 10
a = 0
b = 1

list_new = []

for i in range(n):
    print(a, end = " ")
    list_new.append(a)
    a, b = b, a + b

print(list_new)
```

```python
# Fibonacci Series - 02

n = int(input("Enter the number of Loops :"))

a = 0
b = 1
print(a, b, end = " ")
for i in range(n - 2):
    c = a + b
    print(c, end = " ")
    a = b
    b = c
```

```python
# Fibonacci Series - 03 [Using While Loop]

n = int(input("Enter the number of Loops :"))

a = 0
b = 1
count = 0

while count < n:
    print(a, end = " ")
    a, b = b, a + b
    count += 1
```

```python
# Fibonacci Series - 04 [Using Recursion]

def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(10):
    print(fibonacci(i), end = " ")
```

```python
# Fibonacci Series - 05 [Using list]

n = 10
fib = [0, 1]

for i in range(2, n):
    fib.append(fib[i - 1] + fib[i - 2])
print(fib[:n])
```

```python
# Fibonacci Series - 06 [Using Yield]

def fibonacci(n):
    a, b = 0, 1

    for i in range(n):
        yield a
        a, b = b, a + b
print(list(fibonacci(10)))
```

```python
# Fibonacci Series - 07 [Using memoization]

from functools import lru_cache

@lru_cache(maxsize=None)
def fibonacci(n):
    if n <= 1:
        return n

    return fibonacci(n - 1) + fibonacci(n - 2)

print([fibonacci(i) for i in range(10)])
```
